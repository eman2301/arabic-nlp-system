# Arabic-to-English Machine Translation

This notebook demonstrates Arabic-to-English machine translation for questions and answers from the AAFAQ dataset.

The translation system uses the OpenRouter API with GPT-4o-mini to translate Arabic text into English.

## 1. Setup and Imports

```python
!pip install openai pandas
```

```python
import pandas as pd
import os
from openai import OpenAI
```

## 2. Load Preprocessed Dataset

```python
df = pd.read_csv('/content/AAFAQ_preprocessed.csv')
```

## 3. Configure Translation Client

```python
client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=os.getenv("OPENROUTER_API_KEY")
)
```

## 4. Arabic-to-English Translation Function

```python
def translate_arabic_to_english(text):

    response = client.chat.completions.create(
        model="openai/gpt-4o-mini",
        messages=[
            {
                "role": "system",
                "content": "You are a professional translator. Translate Arabic text into English only."
            },
            {
                "role": "user",
                "content": f"Translate the following Arabic text into English:\n{text}"
            }
        ],
        max_tokens=1000,
        temperature=0.3
    )

    return response.choices[0].message.content
```

## 5. Translation Examples

```python
for i in range(5):
  question = df['QuestionText'].iloc[i]
  answer = df['Answer'].iloc[i]

  translated_question = translate_arabic_to_english(question)
  translated_answer = translate_arabic_to_english(answer)

  print("Arabic question:", question)
  print("Translated question:", translated_question)
  print("Arabic answer:", answer)
  print("Translated answer:", translated_answer)
  print("--------------------------------------------")
```
